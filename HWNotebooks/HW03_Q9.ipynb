import

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from scarlet_locomotive.model import LinearRegression
```

load data

```python
df = pd.read_csv("https://raw.githubusercontent.com/rahulbhadani/Datasets/master/Hydropower.csv")

x = df["BCR"]
y = df["AnnualProduction"]

df.head()
```

split the data 75% train and 25% test

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

train model

```python
model = LinearRegression(learning_rate=0.1, epochs=1000)
model.fit(x_train, y_train, x_test, y_test)
```

predict Annual Production

```python
new_bcr = [0.5, 1.0, 2.0]
print(model.predict(new_bcr))
```

plot

```python
model.analysis_plot(x_train, y_train)
```
