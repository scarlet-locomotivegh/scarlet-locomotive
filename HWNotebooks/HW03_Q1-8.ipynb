### **3. Constrained Optimization using Lagrange Multiplier**


==================================================

**Step 1:** set up the Lagrangian

$g(x, y, z) = x + 2y + z - 6 = 0$

$\mathcal{L}(x, y, z, \lambda) = x^2 + y^2 + z^2 - \lambda(x + 2y + z - 6)$

==================================================

**Step 2:** derive the system of equations by taking each partial and setting it to zero

$\frac{\partial \mathcal{L}}{\partial x} = 2x - \lambda = 0$

$\frac{\partial \mathcal{L}}{\partial y} = 2y - 2\lambda = 0$

$\frac{\partial \mathcal{L}}{\partial z} = 2z - \lambda = 0$


$\frac{\partial \mathcal{L}}{\partial \lambda} = -(x + 2y + z - 6)$

$x + 2y + z$

$=6$

==================================================

**Step 3:** solve for the critical point

from the first three equations:

$x = \frac{\lambda}{2}, \quad y = \lambda, \quad z = \frac{\lambda}{2}$

substitute into the constraint:

$\frac{\lambda}{2} + 2\lambda + \frac{\lambda}{2} = 6$

$3\lambda = 6$

$\lambda = 2$

so $x = 1, \; y = 2, \; z = 1$


critical point is $(1, 2, 1)$ with $\lambda = 2$

==================================================

**Step 4:** classify the extremum

since $f$ is a bowl shape there is 1 lowest point. Thus $(1, 2, 1)$ is a **global minimum** with minimum value $f = 6$

==================================================

**Step 5:** role of the constraint

without the constraint the smallest value of $f$ would be at $(0, 0, 0)$, but that point does not exists because the constraint forces the answer to stay on the plane $x + 2y + z = 6$, so we are really looking for the point on the plane closest to the origin.

**Step 3:** iterate

### **5. Linear Regression and Regularization**

==================================================

**Step 1:** write in vector form

$L(\mathbf{w}) = (\mathbf{y} - \mathbf{X}\mathbf{w})^\top (\mathbf{y} - \mathbf{X}\mathbf{w}) + \lambda \mathbf{w}^\top \mathbf{w}$

==================================================

**Step 2:** take the gradient and set to zero

$\nabla_{\mathbf{w}} L = -2X^\top y + 2X^\top Xw + 2\lambda w$

$-2X^\top y + 2X^\top Xw + 2\lambda w = 0$

$X^\top Xw + \lambda I w = X^\top y$

$(X^\top X + \lambda I)w = X^\top y$

$(X^\top X + \lambda I)^{-1}X^\top y$ $=w$

==================================================

**Step 3:** compare and impact

original solution: $(X^\top X)^{-1}X^\top y$ $=w$, so the penalty just adds $\lambda I$ before the inverse

the penalty shrinks the weights toward 0, which reduces overfitting and makes the matrix always invertible

### **6. Assumptions in Linear Regression**


==================================================

the simple linear regression model is

$y_i = \beta_0 + \beta_1 x_i + \varepsilon_i$

where $\varepsilon_i$ is the error term

==================================================

**1. Linearity:** the relationship between $x$ and $y$ is a straight line, $y = \beta_0 + \beta_1 x + \varepsilon$

**2. Independence:** each data point's error does not affect any other data point's error

**3. Zero mean:** the errors average out to 0

**4. Constant variance:** the spread of the errors is the same for every value of $x$

**5. Normality:** the errors follow a normal distribution

**6. No relation between $x$ and the errors:** the errors do not depend on $x$, and $x$ is measured without error

**7. $x$ must vary:** the $x$ values can't all be the same, because then the slope can't be found

### **7. Computing Derivative using PyTorch Autograd**

```python
import torch

#r and rs values
r = torch.tensor(30000.1, requires_grad=True)
rs = torch.tensor(0.00887)

#time equation
t = 2 * r * torch.sqrt(r / rs) * torch.sqrt(1 - rs / r)

#find the derivative
dtdr = torch.autograd.grad(t, r)
print("dt/dr =", dtdr[0].item())
```

### **8. Linear Regression in One Variable with Scikit-Learn**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

#read the csv file
df = pd.read_csv("https://raw.githubusercontent.com/rahulbhadani/Datasets/master/RAV4/2020-04-15-05-12-00_2T3Y1RFV8KC014025_CAN_Messages_5_Hz.csv")

#x and y
x = df[["steer_angle"]]
y = df["yaw_rate"]

#split the data
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)

#train the model
model = LinearRegression()
model.fit(x_train, y_train)

#print the coefficients
print("intercept:", model.intercept_)
print("slope:", model.coef_)

#test the model
pred = model.predict(x_test)
print("R2:", r2_score(y_test, pred))
```
